```python
context = gx.get_context()

datasource = context.data_sources.add_pandas(
    name="customer_datasource"
)

data_asset = datasource.add_csv_asset(
    name="customer_data_asset",
    filepath_or_buffer="customer_data.csv"
)

batch_request = data_asset.build_batch_request()

suite = context.suites.add(
    gx.ExpectationSuite(name="customer_data_expectations")
)

validator = context.get_validator(
    batch_request=batch_request,
    expectation_suite=suite
)

validator.expect_column_values_to_not_be_null("customer_id")
validator.expect_column_values_to_be_unique("customer_id")

validator.expect_column_values_to_be_between("age", 0, 120)

validator.expect_column_values_to_match_regex(
    "email",
    r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
)

validator.expect_column_values_to_not_be_null("salary", mostly=0.95)

validator.expect_column_values_to_be_in_set(
    "country",
    ["USA", "Canada", "UK", "Australia"]
)

validator.expect_column_values_to_be_of_type(
    "signup_date",
    "datetime64[ns]"
)

validator.expect_table_row_count_to_be_between(500, 1000)

print("All expectations added successfully (no save needed in ephemeral mode).")
```

```python
import json

context = gx.get_context()

datasource = context.data_sources.add_pandas(
    name="customer_datasource"
)

data_asset = datasource.add_csv_asset(
    name="customer_data_asset",
    filepath_or_buffer="customer_data.csv"
)

batch_request = data_asset.build_batch_request()

suite = context.suites.add(
    gx.ExpectationSuite(name="customer_data_expectations")
)

validator = context.get_validator(
    batch_request=batch_request,
    expectation_suite=suite
)

validator.expect_column_values_to_not_be_null("customer_id")
validator.expect_column_values_to_be_unique("customer_id")

validator.expect_column_values_to_be_between("age", 0, 120)

validator.expect_column_values_to_match_regex(
    "email",
    r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
)

validator.expect_column_values_to_not_be_null("salary", mostly=0.95)

validator.expect_column_values_to_be_in_set(
    "country",
    ["USA", "Canada", "UK", "Australia"]
)

validator.expect_column_values_to_be_of_type(
    "signup_date",
    "datetime64[ns]"
)

validator.expect_table_row_count_to_be_between(500, 1000)

results = validator.validate()

results_json = results.to_json_dict()

issues = []

for res in results_json["results"]:
    exp_name = res["expectation_config"]["type"]
    success = res["success"]
    unexpected_count = res["result"].get("unexpected_count", 0)

    if not success:
        issues.append({
            "expectation": exp_name,
            "unexpected_count": unexpected_count,
            "details": res["result"]
        })

html = """
<html>
<head>
<title>Data Quality Report</title>
<style>
body { font-family: Arial; margin: 40px; }
h1 { color: #333; }
table { border-collapse: collapse; width: 100%; margin-top: 20px; }
th, td { border: 1px solid #ccc; padding: 8px; }
th { background-color: #f2f2f2; }
</style>
</head>
<body>
<h1>Data Quality Report</h1>
<p>This report summarizes validation results from Great Expectations.</p>

<h2>Issues Found</h2>
<table>
<tr><th>Expectation</th><th>Unexpected Count</th></tr>
"""

for issue in issues:
    html += f"<tr><td>{issue['expectation']}</td><td>{issue['unexpected_count']}</td></tr>"

html += """
</table>

<h2>Full Validation JSON</h2>
<pre>
""" + json.dumps(results_json, indent=2) + """
</pre>

</body>
</html>
"""
with open("data_quality_report.html", "w") as f:
    f.write(html)

print("HTML report saved as data_quality_report.html")

print("\nSummary of Issues:")
for issue in issues:
    print(f"- {issue['expectation']}: {issue['unexpected_count']} issues")
```

```python
import pandas as pd
import re

def load_csv(filepath):
    try:
        df = pd.read_csv(filepath)
    except FileNotFoundError:
        raise FileNotFoundError(f"File not found: {filepath}")

    if df.empty:
        raise ValueError("CSV file is empty")

    return df

def clean_phone(phone):
    if phone is None:
        return None

    phone = str(phone).strip()

    digits = re.sub(r"\D", "", phone)

    return digits

def validate_email(email):
    if email is None:
        return False

    email = email.strip()

    pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

    return bool(re.match(pattern, email))
```
